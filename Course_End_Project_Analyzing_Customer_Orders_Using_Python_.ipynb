# 3. Analyze customer orders

## Total Spending

```python
customer_spending = {}

for order in customer_orders:
    customer = order[0]
    price = order[2]

    if customer in customer_spending:
        customer_spending[customer] += price
    else:
        customer_spending[customer] = price

print(customer_spending)
```

## Customer Classification

```python
customer_classification = {}

for customer, total in customer_spending.items():

    if total > 100:
        customer_classification[customer] = "High-value buyer"

    elif total >= 50:
        customer_classification[customer] = "Moderate buyer"

    else:
        customer_classification[customer] = "Low-value buyer"

print(customer_classification)
```

## Final Result

```python
for customer in customer_names:
    print(
        customer,
        "Total Spending:", customer_spending[customer],
        "Classification:", customer_classification[customer]
    )
```

# 4. Generate Business Insights

## Revenue per Product Category

```python
category_revenue = {}

for order in customer_orders:
    category = order[3]
    price = order[2]

    if category in category_revenue:
        category_revenue[category] += price
    else:
        category_revenue[category] = price

print(category_revenue)
```

## Unique Products

```python
unique_products = set()

for order in customer_orders:
    product = order[1]
    unique_products.add(product)

print("Unique Products:")
print(unique_products)
```

## Customers Who Purchased Electronics

```python
electronics_customers = [
    order[0]
    for order in customer_orders
    if order[3] == "Electronics"
]

print("Customers who purchased Electronics:")
print(electronics_customers)
```

## Top 3 Highest-Spending Customers

```python
top_customers = sorted(
    customer_spending.items(),
    key=lambda x: x[1],
    reverse=True
)[:3]

print("Top 3 Highest-Spending Customers:")
print(top_customers)
```

# 5. Organize and Display Data

## Customer Summary

```python
for customer in customer_names:
    print(
        customer,
        "- Total Spending:", customer_spending[customer],
        "- Classification:", customer_classification[customer]
    )
```

## Multiple Categories

```python
customer_categories = {}

for order in customer_orders:
    customer = order[0]
    category = order[3]

    if customer not in customer_categories:
        customer_categories[customer] = set()

    customer_categories[customer].add(category)

multiple_category_customers = [
    customer
    for customer, categories in customer_categories.items()
    if len(categories) > 1
]

print("Customers who purchased from multiple categories:")
print(multiple_category_customers)
```

## Electronics + Clothing

```python
electronics_customers = set()
clothing_customers = set()

for order in customer_orders:
    if order[3] == "Electronics":
        electronics_customers.add(order[0])

    if order[3] == "Clothing":
        clothing_customers.add(order[0])

common_customers = electronics_customers.intersection(clothing_customers)

print("Customers who bought both Electronics and Clothing:")
print(common_customers)
```

#Action 3: Insight Generation

```python
high_value_customers = [
    customer
    for customer, classification in customer_classification.items()
    if classification == "High-value buyer"
]

print("High-Value Customers:")
print(high_value_customers)
```

```python
product_frequency = {}

for order in customer_orders:
    product = order[1]
    product_frequency[product] = product_frequency.get(product, 0) + 1

print("Most Frequently Purchased Products:")

max_frequency = max(product_frequency.values())

for product, frequency in product_frequency.items():
    if frequency == max_frequency:
        print(product, "->", frequency, "purchases")
```

# FINAL RESULT

```python
print("========== FINAL BUSINESS RESULT ==========")

print("\nCustomer Classifications:")
for customer, classification in customer_classification.items():
    print(customer, "->", classification)

print("\nCategory-wise Total Sales:")
for category, revenue in category_revenue.items():
    print(category, "-> $", revenue)

print("\nKey Business Insights:")

highest_revenue_category = max(
    category_revenue,
    key=category_revenue.get
)

highest_spending_customer = max(
    customer_spending,
    key=customer_spending.get
)

high_value_count = sum(
    1 for value in customer_classification.values()
    if value == "High-value buyer"
)

moderate_count = sum(
    1 for value in customer_classification.values()
    if value == "Moderate buyer"
)

low_value_count = sum(
    1 for value in customer_classification.values()
    if value == "Low-value buyer"
)

print("Highest Revenue Category:", highest_revenue_category)

print(
    "Highest Spending Customer:",
    highest_spending_customer,
    "-> $",
    customer_spending[highest_spending_customer]
)

print("High-Value Customers:", high_value_count)
print("Moderate Customers:", moderate_count)
print("Low-Value Customers:", low_value_count)
```
